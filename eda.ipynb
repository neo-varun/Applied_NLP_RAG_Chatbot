```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/amazon_review.csv")

print("Dataset Shape:", df.shape)
print("\nColumns:\n", df.columns)
```

```python
rating_dist = df["overall"].value_counts().sort_index()
print("\nRating Distribution\n", rating_dist)

df["review_length"] = df["reviewText"].fillna("").str.len()

print("\nReview Length Stats:")
print(df["review_length"].describe())

avg_length_by_rating = df.groupby("overall")["review_length"].mean()

print(avg_length_by_rating)
```

Bar plot showing ratings and their counts

```python
plt.figure()
rating_dist.plot(kind="bar")
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
```

Bar plot showing character length and their frequency

```python
plt.figure()
df["review_length"].hist(bins=50)
plt.title("Review Length Distribution")
plt.xlabel("Character Length")
plt.ylabel("Frequency")
```

Box plot showing relationship between ratings and review length

```python
plt.figure()
df.boxplot(column="review_length", by="overall")
plt.title("Review Length by Rating")
plt.suptitle("")
plt.xlabel("Rating")
plt.ylabel("Review Length (Characters)")
```

Bar plot showing ratings and their average length

```python
plt.figure()
avg_length_by_rating.plot(kind="bar")
plt.title("Average Review Length per Rating")
plt.xlabel("Rating")
plt.ylabel("Average Length")
```
